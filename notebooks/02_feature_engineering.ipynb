# 02 Feature Engineering

This notebook demonstrates the end-to-end feature engineering workflow:

- Load processed train, validation, and test datasets
- Apply `CreditDataPreprocessor`
- Apply `CreditFeatureSelector`
- Apply `CreditSampler` on the training set only
- Export the final arrays to `data/processed/` in `.npy` format


```python
from pathlib import Path
import json
import sys

import numpy as np
import pandas as pd

# Automatically locate the project root.
cwd = Path.cwd().resolve()
PROJECT_ROOT = cwd.parent if cwd.name == 'notebooks' else cwd
SRC_DIR = PROJECT_ROOT / 'src'
PROCESSED_DIR = PROJECT_ROOT / 'data' / 'processed'
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from features import CreditDataPreprocessor, CreditFeatureSelector, CreditSampler
```

```python
# Configuration
TARGET_COLUMN = 'preloan_risk_label'
USE_SUBSET_FOR_DEMO = False
TRAIN_ROWS = 120000
VAL_ROWS = 30000
TEST_ROWS = 30000

# Set USE_SUBSET_FOR_DEMO to True when local resources are limited.
print({
    'TARGET_COLUMN': TARGET_COLUMN,
    'USE_SUBSET_FOR_DEMO': USE_SUBSET_FOR_DEMO,
    'TRAIN_ROWS': TRAIN_ROWS,
    'VAL_ROWS': VAL_ROWS,
    'TEST_ROWS': TEST_ROWS,
})
```

```python
# Load processed datasets
train_df = pd.read_csv(PROCESSED_DIR / 'train.csv', low_memory=False)
val_df = pd.read_csv(PROCESSED_DIR / 'val.csv', low_memory=False)
def resolve_split_path(split_name: str) -> Path:
    purified_path = PROCESSED_DIR / f'purified_{split_name}.csv'
    return purified_path if purified_path.exists() else PROCESSED_DIR / f'{split_name}.csv'

train_df = pd.read_csv(resolve_split_path('train'), low_memory=False)
val_df = pd.read_csv(resolve_split_path('val'), low_memory=False)
test_df = pd.read_csv(resolve_split_path('test'), low_memory=False)

if USE_SUBSET_FOR_DEMO:
    train_df = train_df.sample(min(TRAIN_ROWS, len(train_df)), random_state=42).sort_index()
    val_df = val_df.sample(min(VAL_ROWS, len(val_df)), random_state=42).sort_index()
    test_df = test_df.sample(min(TEST_ROWS, len(test_df)), random_state=42).sort_index()

print('train_df shape:', train_df.shape)
print('val_df shape:', val_df.shape)
print('test_df shape:', test_df.shape)
display(train_df.head())
```

```python
# Prepare labels
y_train = train_df[TARGET_COLUMN].astype(int).copy()
y_val = val_df[TARGET_COLUMN].astype(int).copy()
y_test = test_df[TARGET_COLUMN].astype(int).copy()

print('Target distribution in train:')
print(y_train.value_counts().sort_index())
```

```python
# Step 1: Preprocessing
preprocessor = CreditDataPreprocessor(target_column=TARGET_COLUMN)
X_train_processed = preprocessor.fit_transform(train_df)
X_val_processed = preprocessor.transform(val_df)
X_test_processed = preprocessor.transform(test_df)

print('Preprocessed shapes:')
print('X_train_processed:', X_train_processed.shape)
print('X_val_processed:', X_val_processed.shape)
print('X_test_processed:', X_test_processed.shape)

print('Feature group summary:')
print(json.dumps({k: len(v) for k, v in preprocessor.feature_groups_.items()}, indent=2, ensure_ascii=False))
```

```python
# Step 2: Feature selection
selector = CreditFeatureSelector(
    variance_threshold=0.01,
    correlation_threshold=0.8,
    top_k_features=50,
    use_pca=False,
    pca_variance_ratio=0.95,
)

X_train_selected = selector.fit_transform(X_train_processed, y_train)
X_val_selected = selector.transform(X_val_processed)
X_test_selected = selector.transform(X_test_processed)

print('Selected shapes:')
print('X_train_selected:', X_train_selected.shape)
print('X_val_selected:', X_val_selected.shape)
print('X_test_selected:', X_test_selected.shape)
display(X_train_selected.head())
```

```python
# Step 3: Sampling on the training set only
sampler = CreditSampler(random_state=42)
X_train_resampled, y_train_resampled = sampler.fit_resample(X_train_selected, y_train)

X_val = X_val_selected.copy()
X_test = X_test_selected.copy()

print('Final output shapes:')
print('X_train_resampled:', X_train_resampled.shape)
print('y_train_resampled:', y_train_resampled.shape)
print('X_val:', X_val.shape)
print('y_val:', y_val.shape)
print('X_test:', X_test.shape)
print('y_test:', y_test.shape)

print('Resampled target distribution:')
print(y_train_resampled.value_counts().sort_index())
```

```python
# Step 4: Save processed arrays to .npy files
np.save(PROCESSED_DIR / 'X_train_resampled.npy', X_train_resampled.to_numpy(dtype=np.float32))
np.save(PROCESSED_DIR / 'y_train_resampled.npy', y_train_resampled.to_numpy(dtype=np.int32))
np.save(PROCESSED_DIR / 'X_val.npy', X_val.to_numpy(dtype=np.float32))
np.save(PROCESSED_DIR / 'y_val.npy', y_val.to_numpy(dtype=np.int32))
np.save(PROCESSED_DIR / 'X_test.npy', X_test.to_numpy(dtype=np.float32))
np.save(PROCESSED_DIR / 'y_test.npy', y_test.to_numpy(dtype=np.int32))

feature_manifest = {
    'selected_feature_names': X_train_resampled.columns.tolist(),
    'target_column': TARGET_COLUMN,
}
(PROCESSED_DIR / 'feature_manifest.json').write_text(
    json.dumps(feature_manifest, indent=2, ensure_ascii=False),
    encoding='utf-8',
)

print('Saved .npy arrays to:', PROCESSED_DIR)
```
